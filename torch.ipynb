```python
a=torch.tensor([2,4,6,8])
print(a)
#or
b=[2,3,4,5]
torch.tensor(b)
print(a.numel())
print(a.size())
print(a.ndim)
print(a.shape)
print("\n",a.dtype)
```

```python
a=torch.tensor([[18,36,54,78],[36,94,72,81],[24,56,79,3],[1,2,4,5]])
print(a[1])
print(a[2])
print(a[[2,3]])#this is known as fancy index
print(a[[2]])
print(a[[2,3],2])
print(a[[1,2],[1,2]])

```

```python
a=torch.tensor([[18,36,54,78],[36,94,72,81],[24,56,79,3],[1,2,4,5]])
print(a[[0,1,2],[1,2,3]])
print(a[2:4,2:4])
print(a[2:,2:])
print(a[1:3,1:3])
print(a[-3:-1,-3:-1])
print(a[0:4,0:1])
```

```python
# add
# subtract
# divide
# multiply
```

```python
a=torch.tensor([[1,2,3],[4,5,6]])
b=torch.tensor([[7,8,9],[10,11,12]])
print(torch.add(a,b))
print(torch.sum(a))#sum doesnt take two arguments
print(torch.subtract(a,b))
print(torch.multiply(a,b))
print(torch.divide(a,b))
print(torch.mul(a,b))
print(torch.sub(a,b))
print(torch.div(a,b))
```

```python
a=torch.tensor([[1,2,53,4,5,6],[7,8,9,10,11,12]])
print(a)
print(torch.flip(a,dims=[1]))
print(torch.flip(a,dims=[0]))
print(torch.argmax(a))
print(a.argmax())
print(torch.argmin(a))
print(a.argmin())
print(a.max())
print(a.min())
```
